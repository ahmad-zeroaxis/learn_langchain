```python
from dotenv import load_dotenv
import os
import requests
```

```python
# load evironmental variables
load_dotenv()
```

```python
def convert_currency(country_symbol_1: str, country_symbol_2: str, amount: int) -> int:
    """Convert currency"""

    exchangerate_api_key = os.getenv("EXCHANGERATE_API_KEY")
    url=f"https://api.exchangerate.host/convert?access_key={exchangerate_api_key}&from={country_symbol_1}&to={country_symbol_2}&amount={amount}"

    response = requests.get(url)
    return response
```

```python
response = convert_currency("USD", "PKR", 100)
```

```python
print(response.status_code)
```

```python
data = response.json()
```
